```python
import pandas as pd
import numpy as np
import pickle
```

```python
best_rf = pickle.load(open("counterfeit_model.sav", "rb"))
sc = pickle.load(open("scaler.sav", "rb"))
selected_cols_list = pickle.load(open("selected_features.sav", "rb"))

print(selected_cols_list)
```

```python
def is_counterfeit(user_input):
    user_input_scaled = sc.transform(user_input.values)
    prediction = best_rf.predict(user_input_scaled)
    prediction_proba = best_rf.predict_proba(user_input_scaled)
    if prediction[0] == 1:
        print("This product is likely COUNTERFEIT")
    else:
        print("This product is likely GENUINE")
    Counterfeit_Probability = prediction_proba[0][1] * 100
    print("Counterfeit_Probability:", round(Counterfeit_Probability, 2), "%")
```

```python
print("Enter product details:\n")

price = float(input("Price: "))
seller_reviews = int(input("Seller Reviews: "))
product_images = int(input("Number of Product Images: "))
description_length = float(input("Description Length: "))
shipping_time_days = float(input("Shipping Time (days): "))
domain_age_days = int(input("Domain Age (days): "))
views = int(input("Views: "))
purchases = int(input("Purchases: "))
warranty_months = float(input("Warranty (months): "))
brand_suspicious = int(input("Is Brand Suspicious? (1 = Yes, 0 = No): "))

user_input = pd.DataFrame([[price, seller_reviews, product_images, description_length,shipping_time_days, domain_age_days, views, purchases,warranty_months, brand_suspicious]], columns=selected_cols_list)
is_counterfeit(user_input)
```
