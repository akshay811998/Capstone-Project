```python
bool_cols = ['contact_info_complete', 'return_policy_clear', 'bulk_orders','unusual_payment_patterns', 'ip_location_mismatch', 'is_counterfeit']
cft_pdt_data[bool_cols] = cft_pdt_data[bool_cols].astype(int)
```

```python
def quanQual(dataset):
    quan=[]
    qual=[]
    for columnName in cft_pdt_data.columns:
        #print(columnName)
        if(dataset[columnName].dtype=='O'):
            #print("qual")
            qual.append(columnName)
        else:
            #print("quan")
            quan.append(columnName)
    return quan,qual
```

```python
quan,qual=quanQual(cft_pdt_data)
```

```python
from sklearn.impute import SimpleImputer
quan_info=["seller_rating","description_length","shipping_time_days","certification_badges","warranty_months"]
imp = SimpleImputer(missing_values=np.nan, strategy='mean')
cft_pdt_data[quan_info] = imp.fit_transform(cft_pdt_data[quan_info])
```

```python
def is_suspicious_brand(name):
    if not isinstance(name, str) or name.strip() == '':
        return 0
    if re.search(r'\d', name):
        return 1
    if re.search(r'[^A-Za-z0-9\s]', name):
        return 1
    return 0
```

```python
preprocessed['brand_suspicious'] = preprocessed['brand'].apply(is_suspicious_brand)
print(preprocessed.groupby('brand_suspicious')['is_counterfeit'].mean())
```

```python
preprocessed = preprocessed.drop(columns=['brand'])
```

```python
preprocessed.shape
```

```python
from sklearn.preprocessing import LabelEncoder

le_country = LabelEncoder()
le_shipping = LabelEncoder()

preprocessed['seller_country'] = le_country.fit_transform(preprocessed['seller_country'])
preprocessed['shipping_origin'] = le_shipping.fit_transform(preprocessed['shipping_origin'])

preprocessed[['seller_country', 'shipping_origin']].head(10)
```

```python
preprocessed = pd.get_dummies(preprocessed, columns=['category'], dtype=int, drop_first=True)
preprocessed.shape
```

```python
preprocessed.to_csv("counterfeit_products_preprocessed_v2.csv",index=False)
```
